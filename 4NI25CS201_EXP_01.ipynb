```python
import numpy as np
import pandas as pd
print("numpy version:",np.__version__)
print("pandas version:",np.__version__)
marks=np.array([78,85,92,67,88])
print("marks:",marks)
print(type(marks))
print("number of dimensions:",marks.ndim)
print("shape:",marks.shape)
print("number of elements:",marks.size)
print("original:",marks)
print("add 5:",marks+5)
print("multiply by 2:",marks*2)
print("divide by 2:",marks/2)
print("sum:",marks.sum())
print("mean:",marks.mean())
print("maximun:",marks.max())
print("minimum:",marks.min())
print("first marks:",marks[0])
print("second marks:",marks[1])
print("last marks:",marks[-1])
print("index 1 to 3:",marks[1:4])
print("first three:",marks[:3])
print("from index 2:",marks[2:])
print("increment by 2:",marks[::2])
```

```python
student_marks=np.array([
    [75,85,92],[45,56,75],[34,45,76]
])
print(student_marks)
print("shape:",student_marks.shape)
print("mean of all values:",student_marks.mean())
print("column-wise mean:",student_marks.mean(axis=0))
print("row-wise mean:",student_marks.mean(axis=1))
```

```python
data={
    "name":["anu","priya","sita","ram"],
    "branch":["cse","cse","ai/ml","cse"],
    "marks":[34,56,24,56]
}
df=pd.DataFrame(data)
print(df)
print("shape:",df.shape)
print("\ncolumns:",df.columns)
print(df.head())
print(df.tail())
print(df.info())
print(df.describe())
```

```python
print("marks column:")
print(df["marks"])
print("\nbranch and marks:")
print(df[["branch","marks"]])
```

```python
print("first row:")
print(df.iloc[0])
print("first three rows:")
print(df.iloc[:3])
```

```python
print("first 2 rows  and first 2 columns:")
print(df.iloc[0:2,0:2])
```

```python
print("first 3 rows:")
display(df.iloc[0:3])
```

```python
print("\nrows 1 to 3,name and marks:")
display(df.iloc[1:4,0:3])
```

```python
high_scores=df[df["marks"]>40]
high_scores
```

```python
data={
    "name":["asha","bharat","chetan","divya","esha"],
    "branch":["cse","ise","cse","aiml","cse"],
    "marks":[72,91,84,95,68],
    "attendance":[82,94,88,96,75]
}
df=pd.DataFrame(data)
df
display(df["marks"])
display(df["marks"]>80)
display(df[df["marks"]>80])
display(df["attendance"]>85)
display(df[df["attendance"]>85])
display(df.iloc[0:3])
display(df["marks"].max())
display(df["marks"].mean())
display(df[df["branch"]=="cse"])
display(df[(df["marks"]>80) & (df["attendance"]>85)])
```
